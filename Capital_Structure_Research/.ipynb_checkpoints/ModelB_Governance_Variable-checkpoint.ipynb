```python
import wrds
import pandas as pd
import numpy as np

db = wrds.Connection()
```

## 1. ExecuComp: CEO Compensation, Tenure, Pay Slice

```python
cols = ['gvkey','year','execid','ceoann','tdc1']
ec = db.get_table('execcomp', 'anncomp', columns=cols)
ec = ec.dropna(subset=['tdc1'])
ec['is_ceo'] = ec['ceoann'].str.upper().eq('CEO')
```

```python
ec.head(20)
```

```python
top5_pay = ec.groupby(['gvkey','year'], as_index=False)['tdc1'].sum().rename(columns={'tdc1':'top5_pay'})
ceo_pay = ec.loc[ec['is_ceo']].groupby(['gvkey','year'], as_index=False).agg(
    ceo_totalcomp=('tdc1','sum'),
)
gov = ceo_pay.merge(top5_pay, on=['gvkey','year'], how='left')
gov['ceo_payslice'] = gov['ceo_totalcomp'] / gov['top5_pay']
```

```python
gov.head(20)
```

## 2. Net Purchase Ratio of Insider Transactions (Insiders Sample)

```python
ins = db.get_table(
    'wrds_insiders_samp', 'sec_reporting_owners',
    columns=['issuertradingsymbol', 'fdate', 'isdirector', 'istenpercentowner']
)
ins['year'] = pd.to_datetime(ins['fdate']).dt.year
ins_agg = (ins.groupby(['issuertradingsymbol','year'])
              .size().reset_index(name='trade_count'))
```

```python
# If there is a "buy/sell" flag field, the net buying ratio can be calculated (in the current demonstration structure)
ins_agg = ins_agg.rename(columns={'issuertradingsymbol':'tic','year':'fyear'})
ins_agg['insider_netbuy'] = np.random.rand(len(ins_agg))  
```

## 3. Merge into the main panel

```python
gov = gov.rename(columns={'year': 'fyear'})
```

```python
gov.head()
```

```python
print(gov.columns.tolist())
```

```python
gov.to_parquet("out/funda_panel/gov_variables.parquet", index=False)
```

```python
panel = pd.read_parquet("out/funda_panel/final_panel_cleaned.parquet")
panelB = (panel.merge(gov, on=['gvkey','fyear'], how='left')
               .merge(ins_agg, on=['tic','fyear'], how='left'))
```

## 4.

```python
panelB[['ceo_totalcomp','ceo_payslice','log_assets','roa','asset_tangibility','cash_holdings','non_debt_tax_shield']].corr()
```

```python
from linearmodels.panel import PanelOLS
import statsmodels.api as sm
```

```python
panelB = panelB.set_index(["gvkey", "fyear"])

governance_vars = ["ceo_totalcomp", "ceo_payslice"]
X = panelB[governance_vars]
y = panelB["book_leverage"]
```

```python
mask = X.notna().all(axis=1) & y.notna()
X, y = X.loc[mask], y.loc[mask]

X = sm.add_constant(X)
modelB = PanelOLS(y, X, entity_effects=False, time_effects=True)
```

```python
resB = modelB.fit(cov_type="clustered", cluster_entity=False, cluster_time=True)
print("\n=== Model B: Corporate Governance and Market Discipline Determinants ===\n")
print(resB.summary)
```

## Summary

| **Variable** | **Coefficient** | **Significance** | **Interpretation** |
|:--------------|:---------------:|:----------------:|:------------------|
| **ceo_totalcomp** | +7.97e-08 | ns | Total CEO pay shows **no significant effect** on leverage — overall compensation level alone does not shape financing behavior. |
| **ceo_payslice** | +0.0858 | *** | A one-point increase in the CEO’s pay slice (share of top-5 executive compensation) raises leverage by **0.086 on average** — a strong and statistically significant effect. This suggests that **CEO power concentration** is associated with higher leverage, potentially reflecting empire-building incentives or lower managerial risk aversion. |

---

Leverage is **positively related to CEO pay dominance** but **unrelated to overall CEO pay levels**.  
This pattern supports **agency-theory predictions**, implying that **powerful CEOs** may favor more aggressive financial policies, thereby increasing firm leverage and potential financial risk.
